<h1>encodeing the emoji</h1>

```python
def emoji_encode(data):
    li = []
    for i in range(len(data)):
        li.append(repr(data[i].encode())[2:-1])    
    return li
```

```python
df.iloc[:,0] = emoji_encode(df.iloc[:,0])
```

<h2>Lower case</h2>

```python
def lower_casing(val):
    return val.str.lower()
```

```python
df.iloc[:,0] = lower_casing(df.iloc[:,0])
```

<h2>HTML Taging</h2>

```python
import re
def html_taging(b):
    li = []
    for i in range(len(b)):
        li.append(re.sub('<.*?>', "", b[i]))
    return li
```

<h2>urls</h2>

```python
def urls(b):
    li = []
    for i in range(len(b)):
        li.append(re.sub(r'http\S+|www\S+', '', b[i]))
    return li
df.iloc[:,0] = urls(df.iloc[:,0])
```

<h2>punctuations </h2>

```python
def punctiations(b):
    li = []
    for i in range(len(b)):
        li.append(re.sub(r'''[!"#$%&'()*+,\-./:;<=>?@\[\\\]^_`{|}~]''', '', b[i]))
    return li
df.iloc[:,0] = punctiations(df.iloc[:,0])
```

<h1>Spell checker</h1>

```python
spell = SpellChecker()
def spell_checker(data):
    li = []
    for i in range(len(data)):
        words = data[i].split()
        corrected = [
        spell.correction(word) if spell.correction(word) is not None else word
        for word in words 
        ]
        
       # print(f"{i} {corrected}")
        val = ' '.join(corrected)
        li.append(val)
        
spell_checker(df.iloc[:,0])
```

<h1>Removing stop words</h1>

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word not in stop_words:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)

def rm_stp_word(data):
    li = []
    for i in range(len(data)):
        li.append(remove_stopwords(data[i]))
    return li
        
```
